# Importing Neccessary Libraries 

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import time
```

# CONSTANTS: Header and URLs

```python
# List of university URLs
urls = [
    "https://free-apply.com/en/university/1003100014",
    "https://free-apply.com/en/university/1003100025",
    "https://free-apply.com/en/university/1003100003",
    "https://free-apply.com/en/university/1003100019",
    "https://free-apply.com/en/university/1003100002",
    "https://free-apply.com/en/university/1003100004",
    "https://free-apply.com/en/university/1003100020",
    "https://free-apply.com/en/university/1003100022",
    "https://free-apply.com/en/university/1003100021",
    "https://free-apply.com/en/university/1003100010",
    "https://free-apply.com/en/university/1003100001",
    "https://free-apply.com/en/university/1003100008",
    "https://free-apply.com/en/university/1003100012",
    "https://free-apply.com/en/university/1003100006",
]

# Set headers to mimic a browser request
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/114.0.0.0 Safari/537.36",
    "Accept-Language": "en-US,en;q=0.9",
}
```

# Scraping Function


```python

def scrape_courses(url):
    try:
        response = requests.get(url, headers=headers)
        response.raise_for_status()
        soup = BeautifulSoup(response.text, 'html.parser')

        # University details
        university_name = soup.find('h1', class_="text-2xl")
        university_name = university_name.get_text(strip=True) if university_name else "N/A"

        location = soup.find('span', string=lambda x: x and "Azerbaijan" in x)
        location = location.get_text(strip=True) if location else "N/A"

        founded_year = soup.find('span', string=lambda x: x and "Founded" in x)
        founded_year = founded_year.get_text(strip=True).replace("Founded in ", "") if founded_year else "N/A"

        university_id = soup.find('span', string=lambda x: x and "ID" in x)
        university_id = university_id.get_text(strip=True).replace("ID ", "") if university_id else "N/A"

        online_status = soup.find('span', string=lambda x: x and "Online" in x)
        online_status = online_status.get_text(strip=True) if online_status else "N/A"

        admission_status = soup.find('span', string=lambda x: x and "Admissions" in x)
        admission_status = admission_status.get_text(strip=True) if admission_status else "N/A"

        # Extracting course details
        courses = []
        course_sections = soup.find_all('details')
        if not course_sections:
            courses.append({
                "University Name": university_name,
                "Location": location,
                "Founded Year": founded_year,
                "University ID": university_id,
                "Online Status": online_status,
                "Admission Status": admission_status,
                "Course Name": "N/A",
                "Language": "N/A",
                "Mode": "N/A",
                "Fee Domestic": "N/A",
                "Fee International": "N/A",
            })
        else:
            for section in course_sections:
                course_name = section.find('h4').get_text(strip=True) if section.find('h4') else "N/A"
                details = section.find_all('h4', class_='font-semibold text-lg')
                language = details[0].get_text(strip=True) if len(details) > 0 else "N/A"
                mode = details[1].get_text(strip=True) if len(details) > 1 else "N/A"
                fee_domestic = details[3].get_text(strip=True) if len(details) > 2 else "N/A"
                fee_international = details[2].get_text(strip=True) if len(details) > 3 else "N/A"
                courses.append({
                    "University Name": university_name,
                    "Location": location,
                    "Founded Year": founded_year,
                    "University ID": university_id,
                    "Online Status": online_status,
                    "Admission Status": admission_status,
                    "Course Name": course_name,
                    "Language": language,
                    "Mode": mode,
                    "Fee Domestic": fee_domestic,
                    "Fee International": fee_international,
                })

        return courses
    except Exception as e:
        print(f"Error scraping {url}: {e}")
        return []
```

# Main Function 

```python
# Scraping the universities
all_courses = []
for url in urls:
    print(f"Scraping {url}...")
    courses = scrape_courses(url)
    if courses:
        all_courses.extend(courses)
    else:
        print(f"No data scraped from {url}")
    time.sleep(10)  # Delaying to reduce risk of blocking or harming website
```

# RUN

```python
# Convert the list of dictionaries to a DataFrame
df = pd.DataFrame(all_courses)

# Save the DataFrame to a CSV file
df.to_csv("data/scraped_data_team_31.csv", index=False, encoding='utf-8')

print(f"Data saved to scraped_data_team_31.csv")
```
